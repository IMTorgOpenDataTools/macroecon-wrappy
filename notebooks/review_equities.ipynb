An option chain (or option matrix) is a table that lists all available option contracts for a specific underlying security, organized by expiration date and strike price.

The option chain provides a snapshot of crucial information for traders. Key data points typically include: 
* Strike Price: The predetermined price at which the option holder can buy (call) or sell (put) the underlying asset.
* Expiration Date: The date the contract expires. Options with less time until expiration lose value faster due to time decay (theta).
* Calls and Puts: The chain is separated into sections for call options (right to buy) and put options (right to sell).
* Bid and Ask: The highest price a buyer is willing to pay (bid) and the lowest price a seller is willing to accept (ask).
* Last Price: The price at which the most recent trade occurred.
* Volume: The total number of contracts traded during the current day.
* Open Interest (OI): The total number of outstanding contracts that have not yet been closed or exercised.
* Implied Volatility (IV): The market's forecast of the underlying asset's volatility, derived from the option's price.
* "The Greeks": Some platforms include risk measures like Delta, Gamma, Theta, and Vega, which quantify how sensitive the option's price is to various factors.

### Multiple Firms (Tickers)

```python
tickers = firms.index.to_list()[:10]
top_ten = yf.download(tickers, period="max")
```

```python
def separate_batch_call_to_dict_of_dfs(batch_tickers):
    """..."""
    top_lvls = list( set( batch_tickers.columns.get_level_values(0)) )
    symbols = list( set( batch_tickers.columns.get_level_values(1)) )
    dfs = {}
    for symbol in symbols:
        df = pd.DataFrame()
        for idx, lvl in enumerate(top_lvls):
            tmp = batch_tickers.loc[:, (lvl, symbol)]
            df.insert(idx, lvl, tmp)
        dfs[symbol] = df
    return dfs

dfs = separate_batch_call_to_dict_of_dfs(top_ten)
```
