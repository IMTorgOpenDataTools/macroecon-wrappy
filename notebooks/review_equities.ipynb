A company's market weight (or market-cap weight) refers to its proportion of an index, determined by its total market value (market capitalization) compared to the total value of all companies in that index, meaning larger companies naturally have a greater influence on the index's movement. 

```python
firms = tech.top_companies
pct_of_sector = firms['market weight'].sum()
print(f'Percent of sector is represented by sum of market weights: {round(pct_of_sector,4)*100}')
firms.head(5)
```

### Multiple Firms (Tickers)

```python
tickers = firms.index.to_list()[:10]
top_ten = yf.download(tickers, period="max")
```

```python
def separate_batch_call_to_dict_of_dfs(batch_tickers):
    """..."""
    top_lvls = list( set( batch_tickers.columns.get_level_values(0)) )
    symbols = list( set( batch_tickers.columns.get_level_values(1)) )
    dfs = {}
    for symbol in symbols:
        df = pd.DataFrame()
        for idx, lvl in enumerate(top_lvls):
            tmp = batch_tickers.loc[:, (lvl, symbol)]
            df.insert(idx, lvl, tmp)
        dfs[symbol] = df
    return dfs

dfs = separate_batch_call_to_dict_of_dfs(top_ten)
```
